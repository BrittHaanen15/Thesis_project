```python
import matplotlib.pyplot as plt

#Dose_avg LET
kV = [50, 100, 150, 200, 225]
Al = [7.33, 4.41, 3.01, 2.26, 2.02]
Cu = [9.07, 5.3, 3.29, 2.33, 2.03]

plt.plot(kV, Al, label = '2.0 mm Al')
plt.plot(kV, Cu, label = '0.3 mm Cu')
plt.legend()
plt.title('Dose-avg LET as function of tube voltage')
plt.xlabel('Tube voltage (kV)')
plt.ylabel('LET_d (keV/um)')
```

```python
#Track_avg LET
kV = [50, 100, 150, 200, 225]
Al = [3.98, 2.18, 1.47, 1.1, 0.98]
Cu = [4.42, 2.41, 1.54, 1.11, 0.97]

plt.plot(kV, Al, label = '2.0 mm Al')
plt.plot(kV, Cu, label = '0.3 mm Cu')
plt.legend()
plt.title('Track-avg LET as function of tube voltage')
plt.xlabel('Tube voltage (kV)')
plt.ylabel('LET_t (keV/um)')
```
