```python
import random

Letters = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J', 'K', 'L']
random_letter = random.randint(0,11)

if random_letter == 3 or random_letter == 11:
    random_nr = random.randint(1,10)
else:
    random_nr = random.randint(1,16)

print(str(Letters[random_letter]) + str(random_nr))
```
